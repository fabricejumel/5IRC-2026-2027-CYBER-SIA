# 🛡️ Laboratoire d'Attaques Adversariales sur YOLOv8, Fast Gradient Sign Method

### 🎯 Objectif du Notebook
Ce notebook a pour but de démontrer et de comparer l'impact des **attaques adversariales** sur un modèle de vision par ordinateur de pointe (**YOLOv8**).

L'expérimentation se déroule en trois étapes clés sur un jeu de test de 8 images :
1. **Évaluation de référence (Baseline) :** Observation des détections initiales du modèle sur les images propres.
2. **Perturbation par Bruit Blanc Aléatoire :** Application d'un bruit uniforme de même intensité ($\epsilon$) pour tester la robustesse basique du réseau face à une dégradation aléatoire.
3. **Attaque par Gradient (FGSM - Fast Gradient Sign Method) :** Exploitation du gradient de la perte calculée par rapport aux pixels de l'entrée (`requires_grad=True`) pour générer une perturbation malveillante ciblée.

*Hypothèse à valider :* Le bruit blanc aléatoire ne devrait que très peu perturber les prédictions, tandis que l'attaque FGSM (guidée par le gradient) devrait réussir à tromper le modèle ou à faire chuter drastiquement la confiance des boîtes de détection.

```python
!pip install -q ultralytics
import torch
import torchvision
import cv2
import matplotlib.pyplot as plt
from PIL import Image
from ultralytics import YOLO

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Utilisation du périphérique :", device)
```

### 📂 Étape 1 : Chargement et Visualisation du Jeu de Données de Test

Dans cette première étape, nous préparons notre environnement de travail en récupérant un jeu de données d'évaluation léger (`COCO8`) via les utilitaires d'Ultralytics.

Nous extrayons un lot de **8 images de test** que nous affichons sous forme de grille (`2x4`). Cela nous permet d'inspecter visuellement les données d'origine avant de les soumettre au modèle YOLOv8 et d'appliquer nos perturbations.

```python
import glob
import cv2
import matplotlib.pyplot as plt
from PIL import Image
from ultralytics.utils.downloads import download

# 1. Téléchargement du jeu de données d'exemple COCO8
download("https://ultralytics.com/assets/coco8.zip", dir="./datasets")

# 2. Récupération robuste de 8 images minimum
image_paths = sorted(glob.glob("./datasets/coco8/**/*.jpg", recursive=True))
if len(image_paths) < 8:
    image_paths = (image_paths * ((8 // len(image_paths)) + 1))[:8]
else:
    image_paths = image_paths[:8]

print(f"✅ {len(image_paths)} images prêtes pour le test.")

# 3. Affichage direct des 8 images brutes du dataset
fig, axes = plt.subplots(2, 4, figsize=(15, 8))
axes = axes.flatten()

for idx, img_path in enumerate(image_paths):
    raw_img = Image.open(img_path).convert("RGB")
    axes[idx].imshow(raw_img)
    axes[idx].axis("off")
    axes[idx].set_title(f"Dataset Image {idx+1}")

plt.tight_layout()
plt.show()
plt.close(fig)
```

```python
# Chargement du modèle YOLOv8 nano en mode entraînement (nécessaire pour les gradients)
model = YOLO("yolov8n.pt").to(device)
model.model.train()

# Affichage initial des 8 images
fig, axes = plt.subplots(2, 4, figsize=(15, 8))
axes = axes.flatten()

for idx, img_path in enumerate(image_paths):
    results = model(img_path, verbose=False)[0]
    res_plotted = results.plot()
    res_rgb = cv2.cvtColor(res_plotted, cv2.COLOR_BGR2RGB)

    axes[idx].imshow(res_rgb)
    axes[idx].axis("off")
    axes[idx].set_title(f"Image originale {idx+1}")

plt.tight_layout()
plt.show()
```

### ⚡ Étape 2 : Implémentation de l'Attaque Adversariale (FGSM)

Cette étape constitue le cœur méthodologique de l'expérimentation. Nous implémentons la méthode **FGSM (*Fast Gradient Sign Method*)** adaptée à l'architecture de YOLOv8.

#### 🔬 Principe de l'attaque :
1. **Suivi des gradients :** L'image d'entrée est convertie en tenseur PyTorch avec l'option `requires_grad=True` pour autoriser le calcul des gradients par rapport aux pixels.
2. **Propagation avant (*Forward Pass*) :** L'image passe à travers le modèle YOLOv8 pour obtenir les prédictions brutes.
3. **Calcul de la fonction de coût (*Loss*) :** Nous définissons une fonction de perte orientée pour maximiser l'écart sur les tenseurs de sortie du modèle.
4. **Extraction et application du gradient :** Par rétropropagation (`loss.backward()`), nous récupérons le signe du gradient de l'image (`input_tensor.grad.data.sign()`).
5. **Perturbation bornée :** Une modification imperceptible (à l'aide d'un facteur d'intensité $\epsilon$) est ajoutée aux pixels d'origine, puis l'image finale est clippée dans l'intervalle $[0, 1]$ pour conserver des valeurs valides.

```python
import torch.nn.functional as F
from torchvision import transforms

preprocess = transforms.Compose([
    transforms.Resize((640, 640)),
    transforms.ToTensor(),
])

def fgsm_attack_yolo(model, image_path, epsilon=0.05):
    raw_image = Image.open(image_path).convert("RGB")
    original_size = raw_image.size

    input_tensor = preprocess(raw_image).unsqueeze(0).to(device)
    input_tensor.requires_grad = True

    # Passage dans le modèle brut
    preds = model.model(input_tensor)

    # Extraction sécurisée d'un tenseur exploitable pour le gradient (selon la structure renvoyée)
    if isinstance(preds, dict):
        # Si c'est un dictionnaire, on prend la première valeur tensorielle trouvée
        target_tensor = next(v for v in preds.values() if isinstance(v, torch.Tensor))
    elif isinstance(preds, (list, tuple)):
        target_tensor = preds[0] if isinstance(preds[0], torch.Tensor) else preds[0][0]
    else:
        target_tensor = preds

    # Loss orientée sur le tenseur extrait
    loss = -torch.mean(torch.abs(target_tensor))

    model.zero_grad()
    loss.backward()

    data_grad = input_tensor.grad.data
    perturbed_image = input_tensor + epsilon * data_grad.sign()
    perturbed_image = torch.clamp(perturbed_image, 0, 1)

    to_pil = transforms.ToPILImage()
    adv_img_pil = to_pil(perturbed_image.squeeze(0).cpu())
    adv_img_pil = adv_img_pil.resize(original_size)

    return adv_img_pil
```

### 🚀 Étape 3 : Exécution de l'Attaque et Comparaison Visuelle

Dans cette dernière étape, nous mettons en pratique notre pipeline d'attaque sur le lot d'images.


 **L'image modifiée par l'attaque FGSM en fonction de epsilon** (guidée par le gradient des poids du modèle).

Cela permet d'observer en direct le moment précis où l'IA perd ses repères, là où l'œil humain perçoit ou non la perturbation.

```python
epsilon_val = 0.1
  # Intensité de la perturbation

fig, axes = plt.subplots(2, 4, figsize=(15, 8))
axes = axes.flatten()

for idx, img_path in enumerate(image_paths):
    adv_image = fgsm_attack_yolo(model, img_path, epsilon=epsilon_val)

    adv_results = model(adv_image, verbose=False)[0]
    adv_plotted = adv_results.plot()

    adv_rgb = cv2.cvtColor(adv_plotted, cv2.COLOR_BGR2RGB)
    axes[idx].imshow(adv_rgb)
    axes[idx].axis("off")
    axes[idx].set_title(f"Attaquée (eps={epsilon_val})")

plt.tight_layout()
plt.show()
```

### 🚀 Étape 4 : Exécution de l'Attaque et Comparaison Visuelle

Dans cette dernière étape, nous mettons en pratique notre pipeline d'attaque sur le lot d'images.

 Nous comparons côte à côte :
1. **L'image originale** et ses détections de référence.
2. **L'image corrompue par un bruit blanc aléatoire** (pour tester la robustesse basique du modèle face au hasard).
3. **L'image modifiée par l'attaque FGSM** (guidée par le gradient des poids du modèle).

Cela permet d'observer en direct le moment précis où l'IA perd ses repères, là où l'œil humain perçoit ou non la perturbation.

```python
epsilon_val = 0.1  # Intensité de la perturbation
num_samples = min(4, len(image_paths))  # On prend 4 images pour une comparaison visuelle claire 3x4

fig, axes = plt.subplots(3, num_samples, figsize=(16, 12))

for idx in range(num_samples):
    img_path = image_paths[idx]
    raw_image = Image.open(img_path).convert("RGB")
    original_size = raw_image.size

    # --- 1. Original ---
    orig_results = model(raw_image, verbose=False)[0]
    axes[0, idx].imshow(cv2.cvtColor(orig_results.plot(), cv2.COLOR_BGR2RGB))
    axes[0, idx].axis("off")
    if idx == 0: axes[0, idx].set_ylabel("Originale", fontsize=12, fontweight='bold')

    # --- 2. Bruit Blanc (Random Noise) ---
    input_tensor = preprocess(raw_image).unsqueeze(0).to(device)
    noise = torch.randn_like(input_tensor) * epsilon_val
    white_noise_tensor = torch.clamp(input_tensor + noise, 0, 1)

    white_pil = transforms.ToPILImage()(white_noise_tensor.squeeze(0).cpu()).resize(original_size)
    white_results = model(white_pil, verbose=False)[0]
    axes[1, idx].imshow(cv2.cvtColor(white_results.plot(), cv2.COLOR_BGR2RGB))
    axes[1, idx].axis("off")
    if idx == 0: axes[1, idx].set_ylabel(f"Bruit Blanc (eps={epsilon_val})", fontsize=12, fontweight='bold')

    # --- 3. Attaque FGSM (Gradient calculé) ---
    adv_image = fgsm_attack_yolo(model, img_path, epsilon=epsilon_val)
    adv_results = model(adv_image, verbose=False)[0]
    axes[2, idx].imshow(cv2.cvtColor(adv_results.plot(), cv2.COLOR_BGR2RGB))
    axes[2, idx].axis("off")
    if idx == 0: axes[2, idx].set_ylabel(f"Attaque FGSM (eps={epsilon_val})", fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
plt.close(fig)
```

### 🚀 Étape 5 : Exécution de l'Attaque et Comparaison Visuelle

Dans cette dernière étape, nous mettons en pratique notre pipeline d'attaque sur le lot d'images.

À l'aide d'un curseur interactif (ou d'une valeur fixe d'intensité $\epsilon$), nous comparons côte à côte :
1. **L'image originale** et ses détections de référence.
2. **L'image corrompue par un bruit blanc aléatoire** (pour tester la robustesse basique du modèle face au hasard).
3. **L'image modifiée par l'attaque FGSM** (guidée par le gradient des poids du modèle).

Cela permet d'observer en direct le moment précis où l'IA perd ses repères, là où l'œil humain perçoit ou non la perturbation.

```python
import ipywidgets as widgets
from IPython.display import display, HTML

# 1. Définition de la fonction interactive liée au curseur
def run_attack_with_slider(epsilon_val):
    num_samples = min(4, len(image_paths))
    fig, axes = plt.subplots(3, num_samples, figsize=(16, 12))

    for idx in range(num_samples):
        img_path = image_paths[idx]
        raw_image = Image.open(img_path).convert("RGB")
        original_size = raw_image.size

        # --- 1. Original ---
        orig_results = model(raw_image, verbose=False)[0]
        axes[0, idx].imshow(cv2.cvtColor(orig_results.plot(), cv2.COLOR_BGR2RGB))
        axes[0, idx].axis("off")
        if idx == 0: axes[0, idx].set_ylabel("Originale", fontsize=12, fontweight='bold')

        # --- 2. Bruit Blanc (Random Noise) ---
        input_tensor = preprocess(raw_image).unsqueeze(0).to(device)
        noise = torch.randn_like(input_tensor) * epsilon_val
        white_noise_tensor = torch.clamp(input_tensor + noise, 0, 1)

        white_pil = transforms.ToPILImage()(white_noise_tensor.squeeze(0).cpu()).resize(original_size)
        white_results = model(white_pil, verbose=False)[0]
        axes[1, idx].imshow(cv2.cvtColor(white_results.plot(), cv2.COLOR_BGR2RGB))
        axes[1, idx].axis("off")
        if idx == 0: axes[1, idx].set_ylabel(f"Bruit Blanc\n(eps={epsilon_val:.2f})", fontsize=11, fontweight='bold')

        # --- 3. Attaque FGSM (Gradient calculé) ---
        adv_image = fgsm_attack_yolo(model, img_path, epsilon=epsilon_val)
        adv_results = model(adv_image, verbose=False)[0]
        axes[2, idx].imshow(cv2.cvtColor(adv_results.plot(), cv2.COLOR_BGR2RGB))
        axes[2, idx].axis("off")
        if idx == 0: axes[2, idx].set_ylabel(f"Attaque FGSM\n(eps={epsilon_val:.2f})", fontsize=11, fontweight='bold')

    plt.tight_layout()
    plt.show()
    plt.close(fig)

# 2. Création de l'interface visuelle et du curseur
display(HTML("<h3 style='color: #2b5b84;'>🎛️ Comparatif Interactif : Bruit Blanc vs Attaque FGSM</h3>"))

epsilon_slider = widgets.FloatSlider(
    min=0.0,
    max=1,           # Vous pouvez monter jusqu'à 0.5 pour tester les fortes intensités !
    step=0.01,
    value=0.05,        # Valeur de départ idéale pour voir l'effet subtil
    description='Epsilon (ε):',
    style={'description_width': 'initial', 'handle_color': '#1f77b4'},
    layout=widgets.Layout(width='450px')
)

out = widgets.interactive(run_attack_with_slider, epsilon_val=epsilon_slider)
display(out)
```
